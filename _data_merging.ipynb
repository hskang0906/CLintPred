```python
import pandas as pd

# 데이터 불러오기
clint_data = pd.read_csv("./dataset/k_melloddy/preprocessed_clint.csv")
admet_data = pd.read_csv("./dataset/k_melloddy/preprocessed_DIDBset.csv")

# 붙일 컬럼 지정
admet_columns = ["SMILES", "logP", "pKa", "solubility", "permeability", "plasma_protein_binding", "fu_in_vitro"]

# admet_data에서 필요한 컬럼만 추출
admet_selected = admet_data[admet_columns]

# SMILES 기준으로 clint_data에 admet_selected의 컬럼 병합 (오른쪽에 붙이기)
merged_data = pd.merge(clint_data, admet_selected, on="SMILES", how="left", suffixes=('', '_admet'))

# 결과를 output_file_name에 저장
output_file_name = "./dataset/k_melloddy/test.csv"
merged_data.to_csv(output_file_name, index=False)
```

```python
load_file_name = "./dataset/k_melloddy/test.csv"
output_file = "./dataset/k_melloddy/all_features.csv"
columns = ["Clint", "logP", "pKa", "solubility", "permeability", "plasma_protein_binding"]

# 파일 불러오기
df = pd.read_csv(load_file_name)

# columns의 모든 값이 결측치가 아닌 행만 추출
filtered_df = df.dropna(subset=columns)

# plasma_protein_binding을 PPB라고 했을 때, fu_in_vitro 값을 1-(PPB/100.0)으로 수정
filtered_df['fu_in_vitro'] = 1 - (filtered_df['plasma_protein_binding'].astype(float) / 100.0)

# 결과 출력
print(filtered_df)
```
